```python
try:
    n = int(input())
    if n<0:
        raise ValueError("Invaid Input: Negative Number")
    for i in range(1,n+1):
        spaces = "  " * (n-i)
        hollow = "  " * ((i-1)*2)
        if i==1:
            print(f"{spaces}*")
            continue
        print(f"{spaces}*{hollow}*")
    for i in range(n-1,0,-1):
        spaces = "  " * (n-i)
        hollow = "  " * ((i-1)*2)
        if i==1:
            print(f"{spaces}*")
            continue
        print(f"{spaces}*{hollow}*")
except ValueError:
    print("Invalid Input")
except Exception as e:
    print(e)
```

```python
def build_borders(j, token):
    border = []
    while j>0:
        border.append(f"{token} ")
        j -= 1
        token -= 1
    return border

try:
    n = int(input())
    if n<0:
        raise ValueError("Invaid Input: Negative Number")
    cache = []
    for i in range(n):
        borders = build_borders(i, n)
        repeating_center = f"{n-i} " * (2*(n-i) - 1)
        left = "".join(borders)
        right = "".join(borders[::-1])
        if i!= (n-1):
            cache.insert(0,f"{left}{repeating_center}{right}")
        print(f"{left}{repeating_center}{right}")
    for i in cache:
        print(i)
except ValueError:
    print("Invalid Input")
except Exception as e:
    print(e)
```
